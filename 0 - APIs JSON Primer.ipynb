## 3. Level 3: Lists inside Dictionaries (The API Standard)
This is the most common pattern in APIs. 
You usually get an Object `{}` that contains a List `[]` of items.

**Rule of Thumb:**
* If it's a Dictionary `{}` -> Use a **Key String** `['key']`
* If it's a List `[]` -> Use an **Integer Index** `[0]`

```python
chat_history = {
    "session_id": "abc-123",
    "messages": [
        {"role": "user", "content": "Hello AI"},          # Index 0
        {"role": "assistant", "content": "Hello Human"},  # Index 1
        {"role": "user", "content": "Write me a poem!"} # Index 2
    ]
}

# 1. Get the list of messages
all_messages = chat_history['messages']

# 2. Get the SECOND message (Index 1)
ai_reply = all_messages[1]

# 3. Get the CONTENT of that message
print(ai_reply['content']) 

# Doing it in one line:
print(chat_history['messages'][1]['content'])
```

Task: Get the list of messages and assign it to a variable called all_messages

```python
# EX

print(ai_reply['content'])
```

## 5. Iterating through JSON
Sometimes you don't want just one item; you want to loop through them all.
We can combine `for` loops with our dictionary navigation.

```python
print(f"Forecast for {weather_data['location']}:\n")

for day_data in weather_data['forecast']:
    day_name = day_data['day']
    condition = day_data['conditions']
    # Calculate average temp from the list of temps
    avg_temp = sum(day_data['temps']) / len(day_data['temps'])
    
    print(f"On {day_name}, it will be {condition} with an avg temp of {avg_temp} degrees.")
```

Below is a mock conversation with a Generative AI model.

**Task:** Write a loop that prints the role that said it followed by the content. 

E.g.

user: Hello! 

assistant: Hello! How can I help you today?

```python
chat_response = {
    "id": "chatcmpl-890",
    "object": "chat.completion",
    "created": 1677852288,
    "model": "gpt-training-model-v1",
    "choices": [
        {
            "index": 0,
            "message": {
                "role": "user",
                "content": "Hey, I'm trying to learn about data structures. What exactly is JSON?"
            },
            "finish_reason": "stop"
        },
        {
            "index": 1,
            "message": {
                "role": "assistant",
                "content": "JSON stands for JavaScript Object Notation. Think of it as a lightweight format for storing and transporting data. It's easy for humans to read and easy for machines to parse."
            },
            "finish_reason": "stop"
        },
        {
            "index": 2,
            "message": {
                "role": "user",
                "content": "Okay, that sounds useful. Is it only for JavaScript?"
            },
            "finish_reason": "stop"
        },
        {
            "index": 3,
            "message": {
                "role": "assistant",
                "content": "Not at all! Despite the name, it's language-independent. You can use it with Python, Java, C++, and pretty much any other modern programming language."
            },
            "finish_reason": "stop"
        },
        {
            "index": 4,
            "message": {
                "role": "user",
                "content": "Awesome. Thanks for the info!"
            },
            "finish_reason": "stop"
        }
    ]
}
```
